```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error

X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
y = np.array([2, 5, 10, 17, 26, 37])

# Linear Regression
linear = LinearRegression()
linear.fit(X, y)
linear_pred = linear.predict(X)

# Polynomial Regression
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)
polynomial = LinearRegression()
polynomial.fit(X_poly, y)
poly_pred = polynomial.predict(X_poly)

print("Linear Regression MSE:",
      round(mean_squared_error(y, linear_pred), 2))
print("Polynomial Regression MSE:",
      round(mean_squared_error(y, poly_pred), 2))

X_smooth = np.linspace(1, 6, 100).reshape(-1, 1)

plt.scatter(X, y, label="Actual Data")
plt.plot(X_smooth, linear.predict(X_smooth),
         label="Linear Regression")
plt.plot(X_smooth, polynomial.predict(poly.transform(X_smooth)),
         label="Polynomial Regression")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear vs Polynomial Regression")
plt.legend()
plt.show()
```
