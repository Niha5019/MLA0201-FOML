```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Month numbers
X = [[1], [2], [3], [4], [5], [6]]

# Sales values
y = [100, 120, 140, 160, 180, 200]

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict next 3 months
future = [[7], [8], [9]]
prediction = model.predict(future)

# Display output
print("Future Sales Prediction:")

for i in range(3):
    print("Month", future[i][0], ":", prediction[i])

    # Graph
plt.figure(figsize=(8, 5))

plt.plot(X, y, marker='o', label="Actual Sales")
plt.plot(future, prediction, marker='o',
             linestyle='--', label="Future Sales")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Future Sales Prediction")

plt.legend()
plt.grid()
plt.show()
```
