```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

X = np.array([
    [1, 2], [1.5, 1.8], [2, 2.5],
    [8, 8], [8.5, 9], [9, 8.5]
])

model = GaussianMixture(
    n_components=2, random_state=42, n_init=5
)
model.fit(X)
labels = model.predict(X)

print("Expectation-Maximization Algorithm")
print("Cluster Labels:", labels)
print("Cluster Means:")
print(np.round(model.means_, 2))

plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(
    model.means_[:, 0], model.means_[:, 1],
    marker="X", s=200, label="Cluster Centers"
)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("EM Algorithm Clustering")
plt.legend()
plt.show()
```
