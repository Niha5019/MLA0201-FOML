```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.linear_model import Perceptron

iris = load_iris()

# Use only two classes and two features for visualization
X = iris.data[:100, :2]
y = iris.target[:100]

model = Perceptron(max_iter=1000, random_state=0)
model.fit(X, y)

prediction = model.predict([[5.0, 3.0]])

print("Predicted Class:",
      iris.target_names[prediction[0]])

# Graph
plt.figure(figsize=(7, 5))

plt.scatter(X[y == 0, 0], X[y == 0, 1],
            label="Setosa")

plt.scatter(X[y == 1, 0], X[y == 1, 1],
            label="Versicolor")

plt.scatter(5.0, 3.0, marker='X', s=200,
            label="New Flower")

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Perceptron Iris Classification")
plt.legend()
plt.grid()
plt.show()
```
