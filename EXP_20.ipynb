```python
import matplotlib.pyplot as plt
from sklearn.naive_bayes import GaussianNB

# Dataset: Income, Credit Score
X = [[30000, 550], [40000, 600], [50000, 700],
     [60000, 750], [25000, 500], [70000, 800]]

     # 0 = Rejected, 1 = Approved
y = [0, 0, 1, 1, 0, 1]

     # Train model
model = GaussianNB()
model.fit(X, y)

     # Predict loan approval
prediction = model.predict([[55000, 720]])

if prediction[0] == 1:
   print("Loan Approved")
else:
   print("Loan Rejected")

             # Graph
plt.bar(["Rejected", "Approved"],
[sum(y.count(i) for i in [0]), y.count(1)])
plt.title("Bank Loan Prediction")
plt.ylabel("Number of Customers")
plt.show()
```
