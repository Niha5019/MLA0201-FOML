```python
import matplotlib.pyplot as plt
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix, accuracy_score

# Income and credit score
X = [
    [30000, 550], [35000, 600], [40000, 650],
    [70000, 750], [80000, 800], [90000, 850],
    [25000, 500], [60000, 700]
]

y = ['No', 'No', 'No', 'Yes',
     'Yes', 'Yes', 'No', 'Yes']

model = GaussianNB()
model.fit(X, y)

# Test data
X_test = [[75000, 780], [30000, 520],
          [65000, 720], [20000, 450]]

y_test = ['Yes', 'No', 'Yes', 'No']

prediction = model.predict(X_test)

print("Predictions:", prediction)
print("Accuracy:", accuracy_score(y_test, prediction))

cm = confusion_matrix(y_test, prediction, labels=['No', 'Yes'])

print("Confusion Matrix:")
print(cm)

# Graph
plt.figure(figsize=(6, 5))
plt.imshow(cm, cmap='Blues')
plt.colorbar()

plt.xticks([0, 1], ['No', 'Yes'])
plt.yticks([0, 1], ['No', 'Yes'])

for i in range(2):
    for j in range(2):
        plt.text(j, i, cm[i, j],
                 ha='center', va='center', fontsize=14)

plt.xlabel("Predicted Loan Status")
plt.ylabel("Actual Loan Status")
plt.title("Bank Loan Prediction - Confusion Matrix")
plt.show()
```
