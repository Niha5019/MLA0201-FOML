```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier

# RAM (GB), Battery (mAh)
X = [
    [2, 2000], [3, 2500], [4, 3000],
    [6, 4000], [8, 4500], [8, 5000],
    [12, 5000], [16, 6000]
]

y = ['Low', 'Low', 'Low', 'Medium',
     'Medium', 'Medium', 'High', 'High']

model = DecisionTreeClassifier(random_state=0)
model.fit(X, y)

test = [[8, 4500]]
prediction = model.predict(test)

print("Mobile Price Category:", prediction[0])

# Graph
plt.figure(figsize=(7, 5))

for label in ['Low', 'Medium', 'High']:
    points = [X[i] for i in range(len(X)) if y[i] == label]
    if points:
        plt.scatter([p[0] for p in points],
                    [p[1] for p in points],
                    label=label, s=100)

plt.scatter(8, 4500, marker='X', s=200, label="Test Mobile")

plt.xlabel("RAM (GB)")
plt.ylabel("Battery (mAh)")
plt.title("Mobile Price Classification")
plt.legend()
plt.grid()
plt.show()
```
