```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Car age and price in lakhs
X = [[1], [2], [3], [4], [5]]
y = [10, 8, 6, 4, 2]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[3]])

print("Predicted Car Price:", prediction[0], "Lakhs")

# Graph
plt.figure(figsize=(7, 5))
plt.scatter(X, y, label="Actual Car Prices")
plt.plot(X, model.predict(X), label="Prediction Line")

plt.scatter([3], prediction, marker='X', s=200, label="Predicted Price")

plt.xlabel("Car Age (Years)")
plt.ylabel("Price (Lakhs)")
plt.title("Car Price Prediction")
plt.legend()
plt.grid()
plt.show()
```
