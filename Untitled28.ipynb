```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree

# Data: Income, Loan Amount, Previous Payment
X = [
    [50000, 10000, 1],
    [60000, 15000, 1],
    [25000, 20000, 0],
    [30000, 25000, 0],
    [80000, 10000, 1],
    [20000, 30000, 0]
]

y = ['Good', 'Good', 'Bad', 'Bad', 'Good', 'Bad']

model = DecisionTreeClassifier(random_state=0)
model.fit(X, y)

test = [[55000, 12000, 1]]
prediction = model.predict(test)

print("Credit Score Classification:", prediction[0])

# Graph
plt.figure(figsize=(8, 5))
plot_tree(
    model,
    feature_names=['Income', 'Loan Amount', 'Previous Payment'],
    class_names=model.classes_,
    filled=True
)
plt.title("Credit Score Classification Tree")
plt.show()
```
